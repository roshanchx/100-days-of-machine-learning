```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.info()
```

```python
# FOR CATEGORICAL DATA
```

```python
df['Survived'].value_counts().plot(kind = 'bar')
#plots the bar graph of number of passengers who survived and who didn't
```

```python
#same can be done using a piechart
df['Survived'].value_counts().plot( kind='pie', autopct='%.2f%%')
#we can also add percentages in bar graph using "autopct"
```

```python
# FOR NUMERICAL DATA
```

```python
#we can also make the numerical data to behave as categorical data
```

```python
#Histogram
import matplotlib.pyplot as plt
```

```python
#plt.hist(df['Age'])
#we can also change the no. of bins using bins parameter
plt.hist(df['Age'], bins = 10)
```

```python
# one more kinf of histogram is distplot (this gives a KDE curve)
sns.distplot(df['Age'])
```

```python
#boxplot (gives 5 no. summary)
sns.boxplot(x = df['Fare'])
plt.show()
```
